# Task 1: Exploring and Visualizing the Iris Dataset

## Introduction
This notebook explores the famous **Iris dataset**, which contains measurements of 150 iris flowers from three species: *Setosa*, *Versicolor*, and *Virginica*. The goal is to understand the data through summary statistics and visualizations.

## Problem Statement
We will load, inspect, and visualize the Iris dataset to understand relationships between features like sepal length, sepal width, petal length, and petal width across different species.

```python
# Install required libraries (if not already installed)
# !pip install pandas matplotlib seaborn

# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print('Libraries imported successfully!')
```

## 1. Dataset Understanding and Description
The Iris dataset has 150 rows and 5 columns:
- **sepal_length**: Length of the sepal (cm)
- **sepal_width**: Width of the sepal (cm)
- **petal_length**: Length of the petal (cm)
- **petal_width**: Width of the petal (cm)
- **species**: The flower species (target variable)

```python
# Load the Iris dataset using seaborn
df = sns.load_dataset('iris')

print('Dataset loaded successfully!')
print(f'Shape of dataset: {df.shape}')
```

## 2. Data Loading and Inspection

```python
# Display column names
print('Columns in the dataset:')
print(df.columns.tolist())
```

```python
# Display first 5 rows
print('First 5 rows of the dataset:')
df.head()
```

```python
# Display last 5 rows
print('Last 5 rows of the dataset:')
df.tail()
```

```python
# Basic summary statistics
print('Statistical Summary:')
df.describe()
```

```python
# Check data types and missing values
print('Data Types and Non-Null Counts:')
df.info()
```

```python
# Count of each species
print('Count of each species:')
print(df['species'].value_counts())
```

```python
# Check for missing values
print('Missing values in each column:')
print(df.isnull().sum())
```

## 3. Exploratory Data Analysis (EDA) with Visualizations

```python
# Set the visual style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# --- SCATTER PLOT: Sepal Length vs Sepal Width ---
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width',
    hue='species',
    palette='deep',
    s=100
)
plt.title('Scatter Plot: Sepal Length vs Sepal Width', fontsize=15, fontweight='bold')
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.legend(title='Species')
plt.tight_layout()
plt.savefig('scatter_sepal.png', dpi=150)
plt.show()
print('Scatter plot saved!')
```

```python
# --- SCATTER PLOT: Petal Length vs Petal Width ---
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species',
    palette='Set1',
    s=100
)
plt.title('Scatter Plot: Petal Length vs Petal Width', fontsize=15, fontweight='bold')
plt.xlabel('Petal Length (cm)', fontsize=12)
plt.ylabel('Petal Width (cm)', fontsize=12)
plt.legend(title='Species')
plt.tight_layout()
plt.savefig('scatter_petal.png', dpi=150)
plt.show()
```

```python
# --- HISTOGRAM: Distribution of all features ---
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
colors = ['steelblue', 'salmon', 'mediumseagreen', 'mediumpurple']

for ax, feature, color in zip(axes.flat, features, colors):
    ax.hist(df[feature], bins=20, color=color, edgecolor='black', alpha=0.8)
    ax.set_title(f'Distribution of {feature.replace("_", " ").title()}', fontsize=13)
    ax.set_xlabel(feature.replace('_', ' ').title() + ' (cm)', fontsize=11)
    ax.set_ylabel('Frequency', fontsize=11)
    ax.grid(axis='y', alpha=0.5)

plt.suptitle('Histograms of Iris Features', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('histograms.png', dpi=150)
plt.show()
print('Histogram saved!')
```

```python
# --- HISTOGRAM: Per species ---
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for ax, feature in zip(axes.flat, features):
    for species, group in df.groupby('species'):
        ax.hist(group[feature], bins=15, alpha=0.6, label=species, edgecolor='black')
    ax.set_title(f'{feature.replace("_", " ").title()} by Species', fontsize=13)
    ax.set_xlabel(feature.replace('_', ' ').title() + ' (cm)', fontsize=11)
    ax.set_ylabel('Frequency', fontsize=11)
    ax.legend(title='Species')

plt.suptitle('Feature Distribution by Species', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('histogram_species.png', dpi=150)
plt.show()
```

```python
# --- BOX PLOT: All features ---
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for ax, feature in zip(axes.flat, features):
    sns.boxplot(
        data=df,
        x='species',
        y=feature,
        palette='pastel',
        ax=ax
    )
    ax.set_title(f'Box Plot: {feature.replace("_", " ").title()} by Species', fontsize=13)
    ax.set_xlabel('Species', fontsize=11)
    ax.set_ylabel(feature.replace('_', ' ').title() + ' (cm)', fontsize=11)

plt.suptitle('Box Plots of Iris Features by Species', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('boxplots.png', dpi=150)
plt.show()
print('Box plots saved!')
```

```python
# --- PAIR PLOT: All features ---
plt.figure()
pair_plot = sns.pairplot(df, hue='species', palette='deep', diag_kind='hist', height=2.5)
pair_plot.fig.suptitle('Pair Plot of Iris Dataset', y=1.02, fontsize=15, fontweight='bold')
plt.savefig('pairplot.png', dpi=150, bbox_inches='tight')
plt.show()
print('Pair plot saved!')
```

```python
# --- HEATMAP: Correlation matrix ---
plt.figure(figsize=(8, 6))
# Only numeric columns for correlation
numeric_df = df.select_dtypes(include='number')
correlation_matrix = numeric_df.corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5,
    square=True
)
plt.title('Correlation Heatmap of Iris Features', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig('heatmap.png', dpi=150)
plt.show()
print('Heatmap saved!')
```

## 4. Conclusion and Key Insights

Based on our exploration and visualization of the Iris dataset:

1. **Dataset Structure**: The Iris dataset contains 150 samples with 4 numerical features and 1 categorical target (species). There are **no missing values**.

2. **Species Balance**: Each species (setosa, versicolor, virginica) has exactly **50 samples**, making the dataset perfectly balanced.

3. **Scatter Plot Insights**:
   - Petal length and petal width show the clearest separation between species.
   - *Setosa* is clearly separated from the other two species.
   - *Versicolor* and *Virginica* show some overlap.

4. **Histogram Insights**:
   - Sepal width appears roughly normally distributed.
   - Petal length and petal width show multimodal distributions due to species differences.

5. **Box Plot Insights**:
   - *Setosa* has the smallest petal dimensions but widest sepal width.
   - *Virginica* has the largest petal length and width.
   - Some outliers are visible in sepal width for all species.

6. **Correlation**: Petal length and petal width are highly positively correlated (r ≈ 0.96), making them strong features for species classification.
